```python
import pandas as pd

df = pd.DataFrame({
    "date": [
        "01.01.2023",
        "01.02.2023",
        "01.03.2023",
        "01.04.2023",
        "01.05.2023",
        "01.06.2023"
    ],
    "clients": [1540, 960, 1150, 1230, 1056, 995]
})

depth = 2    # x(n-2), x(n-1)
horizon = 1  # x(n+1)

df_sw = df.copy()
df_sw["x(n-2)"] = df_sw["clients"].shift(2)
df_sw["x(n-1)"] = df_sw["clients"].shift(1)
df_sw["x(n)"]   = df_sw["clients"]
df_sw["x(n+1)"] = df_sw["clients"].shift(-1)

df_sw
```

```python
df_sw_clean = df_sw.dropna().reset_index(drop=True)
df_sw_clean
```

```python
df2 = pd.DataFrame({
    "i": [1, 2, 3, 4, 5, 6],
    "a": [8, 5, 4, 10, 12, 9],
    "b": [111, 112, 106, 114, 96, 116]
})

depth = 2

df_multi = df2.copy()

# создаём лаги
df_multi["a(n-1)"] = df_multi["a"].shift(1)
df_multi["a(n-2)"] = df_multi["a"].shift(2)
df_multi["b(n-1)"] = df_multi["b"].shift(1)
df_multi["b(n-2)"] = df_multi["b"].shift(2)

# удаляем неполные строки
df_multi_clean = df_multi.dropna().reset_index(drop=True)

df_multi_clean
```
