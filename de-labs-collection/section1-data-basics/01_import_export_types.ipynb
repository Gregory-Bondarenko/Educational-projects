# Импорт, экспорт и типы данных

Задание из методички к Deductor Studio: импортировать CSV без заголовков, задать метки столбцам, выгрузить набор в текстовый файл, загрузить обратно и сменить тип `Поле2` на логический. Здесь то же самое сделано в pandas

```python
#Импорт библиотек
import pandas as pd
```

```python
#Импорт файла
df = pd.read_csv("data/input.csv", header=None)
df
```

```python
#Задаем метки к столбцам: Поле1, Поле2, Поле3 и т.д.
df.columns = [f"Поле{i+1}" for i in range(len(df.columns))]
df
```

```python
#Экспортируем набор данных в текстовый файл Project1.txt
df.to_csv("data/Project1.txt", index=False, encoding="utf-8")
```

```python
#Импортируем обратно Project1.txt
df2 = pd.read_csv("data/Project1.txt")
df2
```

```python
#Задаем метки к столбцам: Поле1, Поле2, Поле3 и т.д.
df2.columns = [f"Поле{i+1}" for i in range(len(df2.columns))]
```

```python
#Меняем тип столбца 2 на булево значение
df2["Поле2"] = df2["Поле2"].astype(bool)
df2
```

```python
#Сохраняем
df2.to_csv("data/Project1_final.csv", index=False, encoding="utf-8")
```
