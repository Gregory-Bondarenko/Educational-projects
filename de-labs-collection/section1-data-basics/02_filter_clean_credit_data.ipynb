```python
#Импорт CreditSample.txt
df = pd.read_csv("data/CreditSample.txt", sep="\t", encoding="cp1251")
df.head(3)
```

```python
#Сортировка
df = df.sort_values(by=["Срок ссуды, мес", "Размер ссуды, руб", "Количество иждивенцев"])
df.head(3)
```

```python
#Замена в поле Семейное положение
df["Семейное положение"] = df["Семейное положение"].replace({
    "Да": "Женат/замужем",
    "Нет": "Холост/Не замужем"
})
df["Семейное положение"]
```

```python
#Замена в поле Количество иждивенцев
df["Количество иждивенцев_REPLACE"] = df["Количество иждивенцев"].replace({
    0: "Нет",
    1: "1",
    2: "2 и более",
    3: "2 и более"
})
df["Количество иждивенцев_REPLACE"]
```

```python
#Удаляем старое поле и переименовываем новое
df = df.drop(columns=["Количество иждивенцев"])
df = df.rename(columns={"Количество иждивенцев_REPLACE": "Иждивенцы"})
df["Иждивенцы"]
```

```python
#Фильтр Иждивенцы != Нет
df6 = df[df["Иждивенцы"] != "Нет"]
print("Фильтр (Иждивенцы != Нет):", len(df6))
```

```python
#Фильтр Иждивенцы != н/д (в синтетических данных пропусков нет, фильтр ничего не отсекает)
df7 = df[df["Иждивенцы"] != "н/д"]
print("Фильтр (Иждивенцы != н/д):", len(df7))
```

```python
#Фильтр по условиям (a или b)
df8 = df6[((df6["Размер ссуды, руб"].between(2000, 5000)) & (df6["Цель ссуды"] == "Покупка товара")) |
          (df6["Цель ссуды"] == "Иное")]
print("Фильтр (a или b):", len(df8))
```

```python
#Сортировка по полю Код
df9 = df8.sort_values(by="Код")
df9.head(5)
```

```python
#Сохранение
df9.to_csv("data/Project2_final.csv", index=False, encoding="utf-8")
```
