```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Чтение сырого датасета
df = pd.read_excel("../data/raw_vk_posts.xlsx")
print(f"Размер: {df.shape}")
df.head(5)
```

```python
columns_to_check = ["Просмотры", "Лайки", "Комментарии"]

#Функция удаления выбросов методом IQR
def remove_outliers(df, column):
    Q1 = df[column].quantile(0.25)  #Первый квартиль (25%)
    Q3 = df[column].quantile(0.75)  #Третий квартиль (75%)
    IQR = Q3 - Q1  #Межквартильный размах

    lower_bound = Q1 - 1.5 * IQR  #Нижняя граница
    upper_bound = Q3 + 1.5 * IQR  #Верхняя граница

    filtered_df = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    return filtered_df

#Визуализация выбросов (до удаления)
plt.figure(figsize=(10, 5))
for i, column in enumerate(columns_to_check, 1):
    plt.subplot(1, 3, i)
    sns.boxplot(y=df[column])
    plt.title(f"Выбросы в {column}")

plt.tight_layout()
plt.show()

#Выбросы
for column in columns_to_check:
    df = remove_outliers(df, column)

#Визуализация после удаления выбросов
plt.figure(figsize=(10, 5))
for i, column in enumerate(columns_to_check, 1):
    plt.subplot(1, 3, i)
    sns.boxplot(y=df[column])
    plt.title(f"После удаления выбросов: {column}")

plt.tight_layout()
plt.show()
```

```python
# Сохранение — отдельным файлом, чтобы не терять сырые данные
df.to_excel("../data/vk_posts_clean.xlsx", index=False)
print("Данные очищены и сохранены в ../data/vk_posts_clean.xlsx")
```
