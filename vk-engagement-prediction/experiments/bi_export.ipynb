```python
import pandas as pd
import re

df = pd.read_excel("../data/vk_posts_clean.xlsx")

def count_emojis(text):
    if not isinstance(text, str):
        return 0
    emoji_pattern = re.compile(
        r'[\U0001F600-\U0001F64F\U0001F300-\U0001F5FF\U0001F680-\U0001F6FF\U0001F700-\U0001F77F'
        r'\U0001F780-\U0001F7FF\U0001F800-\U0001F8FF\U0001F900-\U0001F9FF\U0001FA00-\U0001FA6F'
        r'\U0001FA70-\U0001FAFF\U00002702-\U000027B0\U000024C2-\U0001F251]+',
        flags=re.UNICODE
    )
    return len(emoji_pattern.findall(text))

df['Количество символов'] = df['Текст поста'].apply(lambda x: len(str(x)))
df['Количество эмодзи'] = df['Текст поста'].apply(count_emojis)

df.drop('Текст поста', axis=1, inplace=True)

df.to_excel('../data/bi_dataset.xlsx', index=False)
```
